````python
import os
os.makedirs('healthcare_ai/static', exist_ok=True)

html = """<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>AI Healthcare Assistant</title>
  <style>
    * { box-sizing: border-box; margin: 0; padding: 0; }
    body { font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', sans-serif;
           background: #f0f4f8; color: #1a202c; }
    nav { background: #1a56db; color: white; padding: 14px 20px;
          display: flex; align-items: center; gap: 12px;
          position: sticky; top: 0; z-index: 100; box-shadow: 0 2px 8px rgba(0,0,0,.2); }
    nav h1 { font-size: 1.1rem; font-weight: 700; }
    .tabs { display: flex; background: white; border-bottom: 2px solid #e2e8f0; overflow-x: auto; }
    .tab  { flex: 1; min-width: 80px; padding: 12px 8px; text-align: center;
            font-size: .8rem; font-weight: 600; cursor: pointer; color: #718096;
            border-bottom: 3px solid transparent; transition: all .2s; white-space: nowrap; }
    .tab.active { color: #1a56db; border-bottom-color: #1a56db; }
    .tab:hover  { color: #1a56db; background: #ebf4ff; }
    .page { display: none; padding: 16px; max-width: 700px; margin: 0 auto; }
    .page.active { display: block; }
    .card { background: white; border-radius: 12px; padding: 20px;
            box-shadow: 0 1px 6px rgba(0,0,0,.08); margin-bottom: 16px; }
    .card h2 { font-size: 1rem; font-weight: 700; margin-bottom: 14px; color: #1a56db; }
    .stat-grid { display: grid; grid-template-columns: repeat(2, 1fr); gap: 12px; }
    .stat { background: #ebf4ff; border-radius: 10px; padding: 14px; text-align: center; }
    .stat .val { font-size: 1.8rem; font-weight: 800; color: #1a56db; }
    .stat .lbl { font-size: .72rem; color: #4a5568; margin-top: 2px; }
    .form-group { margin-bottom: 12px; }
    label { display: block; font-size: .8rem; font-weight: 600; color: #4a5568; margin-bottom: 4px; }
    input, select { width: 100%; padding: 10px 12px; border: 1.5px solid #e2e8f0;
                    border-radius: 8px; font-size: .9rem; transition: border .2s; }
    input:focus, select:focus { outline: none; border-color: #1a56db; }
    .row2 { display: grid; grid-template-columns: 1fr 1fr; gap: 10px; }
    .btn { width: 100%; padding: 12px; background: #1a56db; color: white;
           border: none; border-radius: 8px; font-size: .95rem; font-weight: 700;
           cursor: pointer; transition: background .2s; margin-top: 4px; }
    .btn:hover { background: #1446b0; }
    .btn:disabled { background: #a0aec0; cursor: not-allowed; }
    .result { margin-top: 14px; padding: 14px; border-radius: 10px;
              font-size: .88rem; line-height: 1.6; display: none; }
    .result.show { display: block; }
    .result.low    { background: #f0fff4; border-left: 4px solid #38a169; }
    .result.medium { background: #fffbeb; border-left: 4px solid #d69e2e; }
    .result.high   { background: #fff5f5; border-left: 4px solid #e53e3e; }
    .result.emergency { background: #fff0f0; border-left: 4px solid #c53030;
                        animation: pulse .8s infinite; }
    @keyframes pulse { 0%,100%{opacity:1} 50%{opacity:.85} }
    .risk-row { margin-bottom: 10px; }
    .risk-label { display: flex; justify-content: space-between;
                  font-size: .82rem; font-weight: 600; margin-bottom: 4px; }
    .bar-track { background: #e2e8f0; border-radius: 6px; height: 10px; overflow: hidden; }
    .bar-fill  { height: 100%; border-radius: 6px; transition: width .8s ease; }
    .bar-low    { background: #38a169; }
    .bar-medium { background: #d69e2e; }
    .bar-high   { background: #e53e3e; }
    #chat-box { height: 320px; overflow-y: auto; background: #f7fafc;
                border-radius: 10px; padding: 12px; margin-bottom: 10px;
                border: 1.5px solid #e2e8f0; }
    .msg { margin-bottom: 10px; display: flex; gap: 8px; align-items: flex-start; }
    .msg.user { flex-direction: row-reverse; }
    .bubble { max-width: 80%; padding: 10px 13px; border-radius: 14px;
              font-size: .87rem; line-height: 1.5; }
    .msg.user .bubble { background: #1a56db; color: white; border-bottom-right-radius: 4px; }
    .msg.bot  .bubble { background: white; border: 1.5px solid #e2e8f0; border-bottom-left-radius: 4px; }
    .avatar { width: 30px; height: 30px; border-radius: 50%; display: flex;
              align-items: center; justify-content: center; font-size: 1rem;
              flex-shrink: 0; background: #ebf4ff; }
    .chat-input { display: flex; gap: 8px; }
    .chat-input input { flex: 1; }
    .send-btn { padding: 10px 16px; background: #1a56db; color: white;
                border: none; border-radius: 8px; cursor: pointer; font-size: 1rem; }
    .sev-badge { display: inline-block; padding: 2px 8px; border-radius: 99px;
                 font-size: .72rem; font-weight: 700; margin-left: 6px; }
    .sev-emergency,.sev-high { background:#fed7d7; color:#c53030; }
    .sev-medium { background:#fefcbf; color:#b7791f; }
    .sev-low    { background:#c6f6d5; color:#276749; }
    .upload-zone { border: 2px dashed #a0aec0; border-radius: 10px; padding: 30px;
                   text-align: center; cursor: pointer; transition: all .2s;
                   background: #f7fafc; margin-bottom: 12px; }
    .upload-zone:hover { border-color: #1a56db; background: #ebf4ff; }
    #preview { max-width: 100%; border-radius: 8px; margin: 10px 0; display: none; }
    table { width: 100%; border-collapse: collapse; font-size: .82rem; }
    th { background: #ebf4ff; padding: 8px 10px; text-align: left; font-weight: 700; color: #1a56db; }
    td { padding: 8px 10px; border-bottom: 1px solid #e2e8f0; }
    .spinner { text-align:center; padding: 20px; color: #718096; font-size:.9rem; }
  </style>
</head>
<body>

<nav>
  <h1>AI Healthcare Assistant</h1>
</nav>

<div class="tabs">
  <div class="tab active" onclick="showPage('dashboard',this)">Dashboard</div>
  <div class="tab" onclick="showPage('risk',this)">Risk Check</div>
  <div class="tab" onclick="showPage('scan',this)">Image Scan</div>
  <div class="tab" onclick="showPage('chat',this)">Chatbot</div>
</div>

<div id="page-dashboard" class="page active">
  <div class="card">
    <h2>Platform Overview</h2>
    <div class="stat-grid" id="stats-grid">
      <div class="spinner" style="grid-column:span 2">Loading stats...</div>
    </div>
  </div>
  <div class="card">
    <h2>Top High-Risk Patients</h2>
    <div id="high-risk-table"><div class="spinner">Loading...</div></div>
  </div>
</div>

<div id="page-risk" class="page">
  <div class="card">
    <h2>Disease Risk Prediction</h2>
    <div class="row2">
      <div class="form-group">
        <label>Age</label>
        <input type="number" id="r-age" placeholder="45" min="1" max="120">
      </div>
      <div class="form-group">
        <label>Gender</label>
        <select id="r-gender"><option>Male</option><option>Female</option></select>
      </div>
    </div>
    <div class="row2">
      <div class="form-group">
        <label>Glucose (mg/dL)</label>
        <input type="number" id="r-glucose" placeholder="120">
      </div>
      <div class="form-group">
        <label>Blood Pressure</label>
        <input type="number" id="r-bp" placeholder="80">
      </div>
    </div>
    <div class="row2">
      <div class="form-group">
        <label>BMI</label>
        <input type="number" id="r-bmi" placeholder="24.5" step="0.1">
      </div>
      <div class="form-group">
        <label>Insulin (uU/mL)</label>
        <input type="number" id="r-insulin" placeholder="80">
      </div>
    </div>
    <div class="form-group">
      <label>Heart Rate (bpm)</label>
      <input type="number" id="r-hr" placeholder="72">
    </div>
    <button class="btn" onclick="predictRisk()">Predict Disease Risk</button>
    <div id="risk-result" class="result"></div>
  </div>
</div>

<div id="page-scan" class="page">
  <div class="card">
    <h2>Medical Image Scanner</h2>
    <div class="upload-zone" onclick="document.getElementById('img-input').click()">
      <p>Click to upload X-ray or skin image</p>
      <p style="font-size:.78rem;color:#a0aec0">PNG or JPG only</p>
    </div>
    <input type="file" id="img-input" accept="image/*" style="display:none"
           onchange="previewImage(event)">
    <img id="preview" alt="Preview">
    <button class="btn" id="scan-btn" onclick="scanImage()" disabled>Analyse Image</button>
    <div id="scan-result" class="result"></div>
  </div>
</div>

<div id="page-chat" class="page">
  <div class="card">
    <h2>Medical Chatbot</h2>
    <div id="chat-box">
      <div class="msg bot">
        <div class="avatar">AI</div>
        <div class="bubble">Hello! I am your AI Healthcare Assistant. Describe your symptoms and I will help assess them. For emergencies always call 911 first.</div>
      </div>
    </div>
    <div class="chat-input">
      <input type="text" id="chat-input" placeholder="Describe your symptoms..."
             onkeydown="if(event.key==='Enter') sendChat()">
      <button class="send-btn" onclick="sendChat()">Send</button>
    </div>
  </div>
</div>

<script>
const API = 'http://localhost:8000';

function showPage(id, el) {
  document.querySelectorAll('.page').forEach(p => p.classList.remove('active'));
  document.querySelectorAll('.tab').forEach(t  => t.classList.remove('active'));
  document.getElementById('page-' + id).classList.add('active');
  el.classList.add('active');
  if (id === 'dashboard') loadDashboard();
}

async function loadDashboard() {
  try {
    const r = await fetch(API + '/dashboard');
    const d = await r.json();
    const s = d.summary;
    document.getElementById('stats-grid').innerHTML = [
      ['total_patients','Patients'],
      ['total_predictions','Predictions'],
      ['total_scans','Image Scans'],
      ['total_chats','Chat Sessions'],
      ['avg_risk_score','Avg Risk Score'],
      ['high_risk_count','High Risk']
    ].map(([k,l]) =>
      '<div class="stat"><div class="val">' + s[k] + '</div><div class="lbl">' + l + '</div></div>'
    ).join('');
    const rows = d.top_high_risk_patients.map(p =>
      '<tr><td>' + p.name + '</td><td>' + p.age + '</td><td>' + p.disease + '</td><td>' + (p.risk_score*100).toFixed(1) + '%</td></tr>'
    ).join('');
    document.getElementById('high-risk-table').innerHTML =
      '<table><tr><th>Name</th><th>Age</th><th>Disease</th><th>Risk</th></tr>' + rows + '</table>';
  } catch(e) {
    document.getElementById('stats-grid').innerHTML =
      '<div style="color:#e53e3e;grid-column:span 2">API offline - start the server first</div>';
  }
}

async function predictRisk() {
  const body = {
    patient_id: 1,
    age:            parseInt(document.getElementById('r-age').value) || 45,
    gender:         document.getElementById('r-gender').value,
    glucose:        parseFloat(document.getElementById('r-glucose').value) || 120,
    blood_pressure: parseFloat(document.getElementById('r-bp').value) || 80,
    bmi:            parseFloat(document.getElementById('r-bmi').value) || 25,
    insulin:        parseFloat(document.getElementById('r-insulin').value) || 80,
    heart_rate:     parseInt(document.getElementById('r-hr').value) || 72
  };
  const el = document.getElementById('risk-result');
  el.className = 'result show medium'; el.innerHTML = 'Analysing vitals...';
  try {
    const r = await fetch(API + '/predict-risk', {
      method:'POST', headers:{'Content-Type':'application/json'}, body: JSON.stringify(body)
    });
    const d = await r.json();
    const maxLevel = d.highest_risk.risk_level;
    el.className = 'result show ' + maxLevel;
    const bars = d.predictions.map(p => {
      const pct = (p.risk_score * 100).toFixed(1);
      return '<div class="risk-row"><div class="risk-label"><span>' + p.disease.replace('_',' ').toUpperCase() +
             '</span><span>' + pct + '% - ' + p.risk_level.toUpperCase() + '</span></div>' +
             '<div class="bar-track"><div class="bar-fill bar-' + p.risk_level +
             '" style="width:' + pct + '%"></div></div></div>';
    }).join('');
    el.innerHTML = '<strong>Risk Assessment</strong><br><br>' + bars +
      '<br><em>AI guidance only - always consult a physician.</em>';
  } catch(e) {
    el.className = 'result show high';
    el.innerHTML = 'Could not connect to API. Make sure server is running.';
  }
}

function previewImage(e) {
  const file = e.target.files[0];
  if (!file) return;
  const reader = new FileReader();
  reader.onload = ev => {
    const img = document.getElementById('preview');
    img.src = ev.target.result; img.style.display = 'block';
    document.getElementById('scan-btn').disabled = false;
  };
  reader.readAsDataURL(file);
}

async function scanImage() {
  const file = document.getElementById('img-input').files[0];
  if (!file) return;
  const el = document.getElementById('scan-result');
  el.className = 'result show medium'; el.innerHTML = 'Scanning image...';
  const fd = new FormData();
  fd.append('file', file);
  try {
    const r = await fetch(API + '/scan-image?patient_id=1', {method:'POST', body: fd});
    const d = await r.json();
    const level = d.diagnosis === 'normal' ? 'low' : 'high';
    el.className = 'result show ' + level;
    const probs = Object.entries(d.all_probabilities).map(([k,v]) => {
      const pct = (v*100).toFixed(1);
      return '<div class="risk-row"><div class="risk-label"><span>' + k.replace('_',' ') +
             '</span><span>' + pct + '%</span></div><div class="bar-track">' +
             '<div class="bar-fill bar-' + (k==='normal'?'low':'high') +
             '" style="width:' + pct + '%"></div></div></div>';
    }).join('');
    el.innerHTML = '<strong>Diagnosis: ' + d.diagnosis.replace('_',' ').toUpperCase() +
      '</strong> (' + (d.confidence*100).toFixed(1) + '% confidence)<br><br>' + probs +
      '<br><em>' + d.recommendation + '</em>';
  } catch(e) {
    el.className = 'result show high';
    el.innerHTML = 'Could not connect to API.';
  }
}

async function sendChat() {
  const input = document.getElementById('chat-input');
  const msg   = input.value.trim();
  if (!msg) return;
  addMessage(msg, 'user');
  input.value = '';
  addMessage('...', 'bot', 'typing');
  try {
    const r = await fetch(API + '/chat', {
      method:'POST', headers:{'Content-Type':'application/json'},
      body: JSON.stringify({patient_id:1, message: msg})
    });
    const d = await r.json();
    document.getElementById('typing')?.remove();
    addMessage(d.response, 'bot', null, d.severity);
  } catch(e) {
    document.getElementById('typing')?.remove();
    addMessage('API offline. Start the server first.', 'bot');
  }
}

function addMessage(text, role, id, severity) {
  const box  = document.getElementById('chat-box');
  const div  = document.createElement('div');
  div.className = 'msg ' + role;
  if (id) div.id = id;
  const badge = severity ?
    '<span class="sev-badge sev-' + severity + '">' + severity + '</span>' : '';
  div.innerHTML =
    '<div class="avatar">' + (role==='user'?'You':'AI') + '</div>' +
    '<div class="bubble">' + text + badge + '</div>';
  box.appendChild(div);
  box.scrollTop = box.scrollHeight;
}

loadDashboard();
</script>
</body>
</html>"""

# KEY FIX: encoding='utf-8' solves the Windows cp1252 error
with open('healthcare_ai/static/index.html', 'w', encoding='utf-8') as f:
    f.write(html)

print('✅ Frontend written to healthcare_ai/static/index.html')
````

## Step 2: Add Static File Serving to FastAPI

````python
# Step 2 fixed for Windows
static_patch = """

# Serve frontend
from fastapi.staticfiles import StaticFiles
from fastapi.responses import FileResponse

app.mount("/static", StaticFiles(directory="healthcare_ai/static"), name="static")

@app.get("/app", include_in_schema=False)
def serve_frontend():
    return FileResponse("healthcare_ai/static/index.html")
"""

# FIX: encoding='utf-8' on BOTH read and write
with open('healthcare_ai/api/main.py', 'r', encoding='utf-8') as f:
    content = f.read()

content = content.replace(
    'if __name__ == "__main__":',
    static_patch + 'if __name__ == "__main__":'
)

with open('healthcare_ai/api/main.py', 'w', encoding='utf-8') as f:
    f.write(content)

print('✅ Static file serving added to API')
print('🌐 Frontend will be available at: http://localhost:8000/app')
````

## Step 4: Create Dockerfile

````python
dockerfile = """FROM python:3.10-slim

RUN apt-get update && apt-get install -y \
    libglib2.0-0 libsm6 libxext6 libxrender-dev libgomp1 curl \
    && rm -rf /var/lib/apt/lists/*

WORKDIR /app

COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

COPY healthcare_ai/ ./healthcare_ai/

EXPOSE 8000

HEALTHCHECK --interval=30s --timeout=10s --start-period=60s \
  CMD curl -f http://localhost:8000/health || exit 1

CMD ["uvicorn", "healthcare_ai.api.main:app", \
     "--host", "0.0.0.0", "--port", "8000", "--workers", "1"]
"""

with open('Dockerfile', 'w', encoding='utf-8') as f:
    f.write(dockerfile)

print('Dockerfile created')
````

## Step 5: Create docker-compose.yml

````python
# Step 5 - docker-compose.yml
compose = """version: '3.9'

services:
  healthcare-api:
    build: .
    container_name: healthcare_ai_app
    ports:
      - "8000:8000"
    volumes:
      - ./healthcare_ai/data:/app/healthcare_ai/data
      - ./healthcare_ai/models:/app/healthcare_ai/models
      - ./healthcare_ai/uploads:/app/healthcare_ai/uploads
    environment:
      - PYTHONUNBUFFERED=1
    restart: unless-stopped
"""

with open('docker-compose.yml', 'w', encoding='utf-8') as f:
    f.write(compose)
print('docker-compose.yml created')
````

## Step 6: GCP Cloud Run Deployment Script

````python
# Step 6 - GCP deploy script
gcp_deploy = """#!/bin/bash
PROJECT_ID="your-gcp-project-id"
SERVICE_NAME="healthcare-ai-platform"
REGION="us-central1"
IMAGE="gcr.io/${PROJECT_ID}/${SERVICE_NAME}"

gcloud auth login
gcloud config set project ${PROJECT_ID}
gcloud services enable cloudbuild.googleapis.com run.googleapis.com
gcloud builds submit --tag ${IMAGE} .
gcloud run deploy ${SERVICE_NAME} \
  --image ${IMAGE} \
  --platform managed \
  --region ${REGION} \
  --allow-unauthenticated \
  --memory 2Gi \
  --cpu 2 \
  --timeout 300 \
  --port 8000

gcloud run services describe ${SERVICE_NAME} \
  --region ${REGION} \
  --format 'value(status.url)'
"""

with open('deploy_gcp.sh', 'w', encoding='utf-8') as f:
    f.write(gcp_deploy)
print('deploy_gcp.sh created')
````

## Step 7: AWS ECS Deployment Script

````python
# Step 7 - AWS deploy script
aws_deploy = """#!/bin/bash
AWS_REGION="us-east-1"
AWS_ACCOUNT_ID=$(aws sts get-caller-identity --query Account --output text)
ECR_REPO="healthcare-ai-platform"
IMAGE_URI="${AWS_ACCOUNT_ID}.dkr.ecr.${AWS_REGION}.amazonaws.com/${ECR_REPO}:latest"
CLUSTER_NAME="healthcare-cluster"

aws ecr create-repository --repository-name ${ECR_REPO} --region ${AWS_REGION} || true

aws ecr get-login-password --region ${AWS_REGION} | \
  docker login --username AWS --password-stdin \
  ${AWS_ACCOUNT_ID}.dkr.ecr.${AWS_REGION}.amazonaws.com

docker build -t ${ECR_REPO} .
docker tag ${ECR_REPO}:latest ${IMAGE_URI}
docker push ${IMAGE_URI}

aws ecs create-cluster --cluster-name ${CLUSTER_NAME} --region ${AWS_REGION} || true

echo "Deployment configured. Complete setup in AWS Console."
"""

with open('deploy_aws.sh', 'w', encoding='utf-8') as f:
    f.write(aws_deploy)
print('deploy_aws.sh created')
````
